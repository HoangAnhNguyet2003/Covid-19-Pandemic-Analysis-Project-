```python
import csv 
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('owid_covid_data.csv')
```

```python
df
```

```python
df['date'] = pd.to_datetime(df['date'])
start_date = '2021-01-01'
end_date = '2023-01-01'

# Lọc dữ liệu theo continent là Asia và khoảng ngày
filtered_df = df[(df['continent'] == 'Asia') & (df['date'] >= start_date) & (df['date'] <= end_date)]

# Loại bỏ các hàng trong filtered_df có new_cases hoặc new_deaths bằng 0 hoặc NaN
df_cleaned = filtered_df[(filtered_df['new_cases'] != 0) & 
                         (filtered_df['new_deaths'] != 0) & 
                         (filtered_df['new_cases'].notna()) & 
                         (filtered_df['new_deaths'].notna())]
print(df_cleaned)
```

#### covid_death

```python
cases_deaths_df = df_cleaned[['continent', 'location', 'date', 'population',
                       'total_cases', 'new_cases', 
                       'total_deaths', 'new_deaths']]

cases_deaths_df.to_csv('data/covid_cases_deaths.csv', index=False)
```

#### covid_vaccin

```python
vaccine_df = df_cleaned[['continent', 'location', 'date', 
                  'total_vaccinations', 'people_vaccinated', 
                  'people_fully_vaccinated', 'new_vaccinations']]

vaccine_df.to_csv('data/covid_vaccine_data.csv', index=False)
```
